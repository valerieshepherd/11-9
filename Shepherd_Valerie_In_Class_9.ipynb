Explain the relationship between the terms "belief," "prior," and "credence"

A belief is a person's judgment or opinion about how likely an event is to occur. A prior is the initial probability assigned to an event before new information or evidence is received. Credence refers to the degree of confidence a person has in that belief. Together, these concepts describe how people form initial expectations about uncertain events and update those expectations when new information becomes available.

# Exercise 5 (8 points)

A hotel group is deciding whether to build a new hotel. Five Decision Advisory Committee (DAC) members each distributed **20 chips** among three outcomes (chip assignment method):

| Stakeholder | No Development | Modest Development | Major Development |
|---|---|---|---|
| CFO | 5 | 7 | 8 |
| Operations SME | 3 | 6 | 11 |
| Market KOL | 2 | 5 | 13 |
| Real-estate SME | 4 | 8 | 8 |
| CEO | 1 | 5 | 14 |

The payoffs (in $M) are 0, 4, and 10 for No, Modest, and Major Development, respectively.

Using Python:

1. Create a DataFrame of the elicited probability distributions (one row per stakeholder) by normalizing the chips, and verify each row sums to 1.
2. Aggregate the distributions into a consensus prior using the equal-weight average (linear opinion pool), and report the standard deviation across stakeholders for each outcome.
3. Compute the expected value (EV) of building the hotel using the consensus prior.
4. New information arrives: *the city council approves the rezoning request*. The likelihoods of approval are 0.10 (No), 0.50 (Modest), and 0.80 (Major). Compute the posterior distribution with Bayes' theorem and the updated EV.
5. Sensitivity analysis: move 10 percentage points of probability from Major to Modest in the consensus prior. How much does the EV change?

```python
import pandas as pd
import numpy as np

data = {
    "Stakeholder": [
        "CFO",
        "Operations SME",
        "Market KOL",
        "Real-estate SME",
        "CEO"
    ],
    "No Development": [5, 3, 2, 4, 1],
    "Modest Development": [7, 6, 5, 8, 5],
    "Major Development": [8, 11, 13, 8, 14]
}

df = pd.DataFrame(data).set_index("Stakeholder")

probabilities = df.div(df.sum(axis=1), axis=0)

print("Individual Probability Distributions:")
print(probabilities)

print("\nRow Sums:")
print(probabilities.sum(axis=1))
```

```python
consensus_prior = probabilities.mean()

std_deviation = probabilities.std()

print("Consensus Prior:")
print(consensus_prior)

print("\nStandard Deviation:")
print(std_deviation)
```

```python
payoffs = np.array([0, 4, 10])

ev = np.dot(consensus_prior.values, payoffs)

print(f"Expected Value of Building the Hotel: ${ev:.2f} million")
```

```python
likelihoods = np.array([0.10, 0.50, 0.80])

unnormalized = consensus_prior.values * likelihoods
posterior = unnormalized / unnormalized.sum()

posterior_ev = np.dot(posterior, payoffs)

print("Posterior Probabilities:")
print(pd.Series(
    posterior,
    index=consensus_prior.index
))

print(f"\nUpdated Expected Value: ${posterior_ev:.2f} million")
```
